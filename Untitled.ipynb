```python
#Installation of required libraries
import numpy as np
import pandas as pd 
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
```

```python
#Reading the dataset
data = load_diabetes()

df = pd.DataFrame(data=data.data,
                           columns=data.feature_names)
```

```python
#inspect data
df.info()
print("\n")

print("\n")
# Add target variable to the DataFrame
df['target'] = data.target



# The first 5 observation units of the data set were accessed.
df.head()

```

```python
#Identifying Outliers
df.describe()

fig, axs = plt.subplots(len(df.columns), 1, figsize=(7, 18), dpi=95)
for i, col in enumerate(df.columns):
    axs[i].boxplot(df[col], vert=False)
    axs[i].set_ylabel(col)
plt.tight_layout()
plt.show()
```

```python

#list of column names
columns = list(df.columns)

clean_df = df.copy()

for name in columns:
    q1, q3 = np.percentile(clean_df[name], [25, 75])
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    clean_df = clean_df[
        (clean_df[name] >= lower) &
        (clean_df[name] <= upper)
    ]

fig, axs = plt.subplots(len(clean_df.columns), 1, figsize=(7, 18), dpi=95)
for i, col in enumerate(clean_df.columns):
    axs[i].boxplot(clean_df[col], vert=False)
    axs[i].set_ylabel(col)
plt.tight_layout()
plt.show()
```
